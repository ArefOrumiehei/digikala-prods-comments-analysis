```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_rows', 100)
plt.rcParams['figure.figsize'] = (12, 6)

# Load products dataset (first 10000 rows due to RAM limitation)
df_products = pd.read_csv('../data/raw/digikala-products.csv', nrows=10000)

print("="*60)
print("PRODUCTS DATASET - BASIC INFO")
print("="*60)
print(f"\nShape: {df_products.shape[0]:,} rows × {df_products.shape[1]} columns")
print(f"\nColumns and Data Types:")
print(df_products.dtypes)
print(f"\n\nMissing Values:")
print(df_products.isnull().sum())
print(f"\n\nMissing Values (%):")
print((df_products.isnull().sum() / len(df_products) * 100).round(2))

print("\n" + "="*60)
print("STATISTICAL SUMMARY")
print("="*60)
print(df_products.describe())

print("\n" + "="*60)
print("UNIQUE VALUES")
print("="*60)
print(f"Unique Categories (Level 1): {df_products['Category1'].nunique()}")
print(f"Unique Categories (Level 2): {df_products['Category2'].nunique()}")
print(f"Unique Brands: {df_products['Brand'].nunique()}")
print(f"Unique Sellers: {df_products['Seller'].nunique()}")

df_products.head(10)
```

```python
# Distribution of ratings
print("\n" + "="*60)
print("RATING DISTRIBUTION")
print("="*60)
print(df_products['Rate'].value_counts().sort_index())

# Distribution of fake products
print("\n" + "="*60)
print("FAKE PRODUCTS DISTRIBUTION")
print("="*60)
print(df_products['Is_Fake'].value_counts())
print(f"\nFake products percentage: {(df_products['Is_Fake'].sum() / len(df_products) * 100):.2f}%")

# Top categories
print("\n" + "="*60)
print("TOP 10 CATEGORIES (LEVEL 1)")
print("="*60)
print(df_products['Category1'].value_counts().head(10))

print("\n" + "="*60)
print("TOP 10 CATEGORIES (LEVEL 2)")
print("="*60)
print(df_products['Category2'].value_counts().head(10))

# Top brands
print("\n" + "="*60)
print("TOP 10 BRANDS")
print("="*60)
print(df_products['Brand'].value_counts().head(10))

# Top sellers
print("\n" + "="*60)
print("TOP 10 SELLERS")
print("="*60)
print(df_products['Seller'].value_counts().head(10))

# Price comparison
print("\n" + "="*60)
print("PRICE ANALYSIS")
print("="*60)
df_products['price_change'] = df_products['Price'] - df_products['min_price_last_month']
df_products['price_change_pct'] = (df_products['price_change'] / df_products['min_price_last_month'] * 100).replace([np.inf, -np.inf], np.nan)

print(f"Products with price increase: {(df_products['price_change'] > 0).sum()}")
print(f"Products with price decrease: {(df_products['price_change'] < 0).sum()}")
print(f"Products with no change: {(df_products['price_change'] == 0).sum()}")
print(f"\nAverage price change: {df_products['price_change'].mean():,.0f} Toman")
print(f"Average price change (%): {df_products['price_change_pct'].mean():.2f}%")
```
